```python
from langchain_openai.embeddings import OpenAIEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_core.documents import Document
```

```python
import config
```

we want to define a question related to data science, use the embadding function and convert to a vector this question, and find another vector similar to the question vector

```python
question = 'what programming languages are more used in data science?'
```

```python
embedding = OpenAIEmbeddings(model='text-embedding-ada-002',api_key=config.api_key)
vectorstore_from_directory = Chroma(
    persist_directory="./intro-to-ds-lectures",
    embedding_function=embedding)
```

```python
retrieved_docs = vectorstore_from_directory.similarity_search(query=question,
                                                                k=5)
```

```python
for i in retrieved_docs:
    print(f"Page Content {i.page_content} \n --------\nLecture Title:{i.metadata['Lecture Title']}\n" )
```
