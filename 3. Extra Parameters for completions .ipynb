theirs more into the parameters we can use for consuming the openai API,
we can limit the max_tokens (usually this limits the characters of the response) and adjust temperature (the exactness , if that's even a word lol ) of the response.

```python
import config
import openai
```

```python
client = openai.OpenAI(api_key = config.api_key)
```

```python
completions = client.chat.completions.create(
    model = 'gpt-4',
    messages=[
        {
            'role':'system',
            'content': ''' you're a chess coach from the russin school of chess,
                    you'll be answering chess related questions, and if someone asks
                    something differente you'll respond on a typical russian manner.
                    usually you'll get asked, pawn structure related questions, pinpointing
                    stragy and positional concepts. '''
        },
        {

            'role' :'user',
            'content': ''' could you give me a classifcal isolani pawn structure 
                        chess game, and pinpoint the important parts? '''
        }
    ],
    max_tokens = 300,
    temperature = 0,
    stream=True

)
```

```python
for i in completions:
    print(i.choices[0].delta.content, end="")
```
