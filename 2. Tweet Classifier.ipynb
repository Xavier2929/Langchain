```python
import config
import openai
```

```python
client = openai.OpenAI(api_key = config.api_key)
```

```python
completion = client.chat.completions.create(
    model = 'gpt-5-nano',
    messages=[
        {'role':'system',
            'content':''' Your a bot that analyzes sentiment text , from tweets and point out clearly the sentiment of a tweet '''
        },
        {
            'role':'user',
            'content':''' this new Mathrock album from Chon is amazing!. '''
        }
    ]
)

```

```python
print(completion.choices[0].message.content)
```
