```python
import config
from langchain_core.runnables import RunnableLambda
from langchain_core.runnables import chain
```

```python
def find_sum(x):
    return sum(x)

def find_square(x):
    return x**2
```

```python
chain_1 = RunnableLambda(find_sum) | RunnableLambda(find_square)
```

```python
chain_1.invoke([2,3])
```

```python
@chain #decorator en python, modifies the function of a function
def runnable_sum(x):
    return sum(x)

@chain
def runnable_square(x):
    return x ** 2
```

```python
type(runnable_sum)
```

```python
chain_2 = runnable_sum | runnable_square
```

```python
chain_2.invoke([1,2,3])
```
